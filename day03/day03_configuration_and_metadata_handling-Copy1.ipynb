Your job as a data scientist is **to make the data usable.**

```python
# These are what usually come from meters / sensors / logs.

energy_readings = [
    {"site_id": "S1", "demand_kw": 120},
    {"site_id": "S2", "demand_kw": 85},
    {"site_id": "S3", "demand_kw": 60},
    {"site_id": "S1", "demand_kw": 150},
    {"site_id": "S4", "demand_kw": 200},
    {"site_id": "S5", "demand_kw": 95},
    {"site_id": "S6", "demand_kw": 40},
    {"site_id": "S3", "demand_kw": 75},
    {"site_id": "S7", "demand_kw": 180},
    {"site_id": "S8", "demand_kw": 110},
]
```

```python
# Metadata usually comes from:  config files , Excel sheets , databases and onboarding forms
site_metadata = {
    "S1": {"location": "Delhi", "type": "commercial", "capacity_kw": 500},
    "S2": {"location": "Bihar", "type": "residential", "capacity_kw": 100},
    "S3": {"location": "Karnataka", "type": "industrial", "capacity_kw": 300},
    "S4": {"location": "Maharashtra", "type": "commercial", "capacity_kw": 600},
    "S5": {"location": "Tamil Nadu", "type": "residential", "capacity_kw": 120},
    "S6": {"location": "Rajasthan", "type": "agricultural", "capacity_kw": 80},
    "S7": {"location": "Gujarat", "type": "industrial", "capacity_kw": 400},
    "S8": {"location": "Punjab", "type": "commercial", "capacity_kw": 350},
    "S9": {"location": "Kerala", "type": "residential", "capacity_kw": 90},
    "S10": {"location": "Odisha", "type": "industrial", "capacity_kw": 250},
    "S11": {"location": "Assam", "type": "commercial", "capacity_kw": 200},
    "S12": {"location": "UP", "type": "residential", "capacity_kw": 110},
    "S13": {"location": "MP", "type": "industrial", "capacity_kw": 280},
    "S14": {"location": "Haryana", "type": "commercial", "capacity_kw": 450},
    "S15": {"location": "Telangana", "type": "agricultural", "capacity_kw": 70},
}
```

# Task 1

```python
def enrich_energy_data(energy_readings ,site_metadata ):
    enriched_data =  []
    
    for reading in energy_readings:
        site_id = reading['site_id']
        meta_data = site_metadata.get(site_id)
    
        if meta_data is None:
            continue
            
        merged = reading.copy()
        merged.update(meta_data)
    
        enriched_data.append(merged)
    
    return enriched_data
        
```

```python
site_data = enrich_energy_data(energy_readings ,site_metadata )
site_data
```

```python
len(site_data)
```

# Task 2

```python
allowed_site_types = {"commercial", "industrial"}
```

```python
def filter_sites_by_type (site_data , allowed_site_types):
    allowed_site_data =[]
    
    for data in site_data:
        if data['type'] in allowed_site_types:
            allowed_site_data.append(data)
            
    return allowed_site_data
```

```python

filtered_site_data = filter_sites_by_type (site_data , allowed_site_types)
filtered_site_data
```

```python
print(len(filtered_site_data))
```

# Task 3 : Aggregate Demand by Location

```python
def aggregate_demand_by_location(filtered_site_data):
    aggregated_data = {}

    for record in filtered_site_data:
        location = record["location"]
        demand = record.get("demand_kw", 0)

        if location not in aggregated_data:
            aggregated_data[location] = 0

        aggregated_data[location] += demand

    return aggregated_data
```

```python
aggregated_demand_by_location (filtered_site_data)
```
