**Design a class-based system to model different low-carbon energy assets, using inheritance to represent shared and specialized behavior.**

1. Base class (parent)

2. Inheritance (child classes)

3. Method overriding

4. Common interfaces

5. Clean object design

### 1️⃣ Base Class — EnergyAsset
Create a base class called EnergyAsset that represents a generic low-carbon energy asset.

```python
class EnergyAsset:
    """
    Base class for energy assets.
    """
    def __init__(self, name , location, capacity_kw):
        self.name  = name
        self.location = location
        self.capacity_kw = capacity_kw
        
    def get_assest_info(self):
        """
        Return  basic asset information
        """
        return {
            'name': self.name,
            'loaction':self.location,
            'capacity_kw' : self.capacity_kw
        }
        
    def estimate_annual_output(self):
        """
        Estimate annual energy or carbon impact.
        
        Must be implemented by child classes.
        """
        raise NotImplementedError(
            "Subclasses must implement estimate_annual_output()"
        )        
```

```python
assest = EnergyAsset("sugarcane biomass" ,"UP", 8)
assest.name
assest.capacity_kw
```

```python
assest.estimate_annual_output()
```

### 2️⃣ Child Class — BiomassPlant

Represents a biomass-based power generation unit.

```python
class BiomassPlant(EnergyAsset):
    """
    Biomass-based power generation plant
    """
    def __init__(self, name, location, capacity_kw, feedstock_type, conversion_efficiency):
        super().__init__(name, location, capacity_kw)
        self.feedstock_type = feedstock_type,
        self.conversion_efficiency = conversion_efficiency

    def estimate_annual_output(self, operating_hours = 7000):
        """
        Estimate annual energy output in kWh.
        """
        return (
            self.capacity_kw*operating_hours *self.conversion_efficiency
        )   # It returns kWh/ year
```

```python
plant = BiomassPlant(name ="Rice Husk Plant", location = "Jamui, Bihar",
                     capacity_kw= 650,feedstock_type="Rice Husk",
                     conversion_efficiency= 0.25 )
print(plant.get_assest_info())
print(f" Annual output: {plant.estimate_annual_output()} kWh")
```

### 3️⃣ Child Class — EnergyStorageSystem

Represents an energy storage system (battery, thermal, etc.).

here --->   capacity_kw → power rating , storage_capacity_kwh → energy content

```python
class EnergyStorageSystem(EnergyAsset):
    """
    Energy storage system (battery , thermal , etc)
    """
    def __init__(self, name, location, capacity_kw, storage_capacity_kwh, round_trip_efficiency):
        super().__init__(name, location, capacity_kw )
        self.storage_capacity_kwh = storage_capacity_kwh
        self.round_trip_efficiency = round_trip_efficiency  # ie. Losses
   
    def estimate_annual_output(self, cycles_per_year = 365):
        """
        Estimate usuable energy deliveried per year (kWh)
        """
        return (
            self.storage_capacity_kwh*cycles_per_year * self.round_trip_efficiency
            )      
```

```python
battery = EnergyStorageSystem(
    name = "Lithium Battery Bank",
    location = "Gujrat",
    capacity_kw = 300,
    storage_capacity_kwh = 1200,
    round_trip_efficiency = 0.81
)


print(battery.get_assest_info())
print(f" Annual usable energy: {battery.estimate_annual_output()} kWh")
```

###  Polymorphism  Example 

```python
assets = [
    BiomassPlant(
        name ="Rice Husk Plant", 
        location = "Jamui, Bihar",
        capacity_kw= 650,
        feedstock_type="Rice Husk",
        conversion_efficiency= 0.25),
    
    EnergyStorageSystem( 
    name = "Lithium Battery Bank",
    location = "Gujrat",
    capacity_kw = 300,
    storage_capacity_kwh = 1200,
    round_trip_efficiency = 0.81),
]
```

```python
for assest in assets:
    info = assest.get_assest_info()
    output = assest.estimate_annual_output()

    print("Asset Info: ",info )
    print("Annual Impact :" ,output)
    print("-"* 70)
    
```
