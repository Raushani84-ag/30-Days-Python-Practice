# 📅 Day 04 — Time, Grouping & Feature Engineering (Real DS Scenario)
🌍 Real-World Scenario (Very Common)

**You’re working on an energy demand forecasting project**.

Raw data comes as **timestamped readings** from multiple sites.
Models don’t learn from raw timestamps — they learn from patterns over time.

**Your job:**  Convert raw time-stamped data into meaningful time-based features.

```python
energy_time_series = [
    {"site_id": "S1", "timestamp": "2024-01-01 08:00", "demand_kw": 120},
    {"site_id": "S1", "timestamp": "2024-01-01 09:00", "demand_kw": 150},
    {"site_id": "S1", "timestamp": "2024-01-01 18:00", "demand_kw": 90},
    {"site_id": "S2", "timestamp": "2024-01-01 08:00", "demand_kw": 60},
    {"site_id": "S2", "timestamp": "2024-01-01 19:00", "demand_kw": 40},
    {"site_id": "S3", "timestamp": "2024-01-01 13:00", "demand_kw": 200},
]
```

# 🔹 Task 1 — Extract Hour from Timestamp
     Rules:
        1.Parse timestamp safely
        2.Do NOT modify original input
        3.Return new list

```python
from datetime import datetime

def add_hour_feature (energy_time_series):
    """
    Add hour feature extracted from timestamp.
    """
    
    hour_extracted_data = []

    for ts in energy_time_series:
        time = ts['timestamp']
        time = datetime.strptime(time , '%Y-%m-%d %H:%M')
        hour = time.hour
        ts['hour'] = hour
        
        hour_extracted_data.append(ts)
        
    return hour_extracted_data      
```

```python
new_energy_time_series = add_hour_feature (energy_time_series)
new_energy_time_series
```

# 🔹 Task 2 — Group Demand by Hour (Feature Engineering)

```python
def aggregate_demand_by_hour(new_energy_time_series):
    
    hourly_aggregated_demand = {}
    for ts in new_energy_time_series:
        hour = ts.get('hour')
        demand = ts['demand_kw']

        if hour not in hourly_aggregated_demand:
            hourly_aggregated_demand[hour] = 0

        hourly_aggregated_demand[hour] += demand

    return hourly_aggregated_demand    
```

```python
aggregating_demand_by_hour(new_energy_time_series)
```

# 🔹 Task 3 — Peak vs Off-Peak Labeling (Business Logic)
    Rule:
    Peak hours → 18 to 22 (inclusive)
    Off-peak → everything else

```python
data_by_hour  = new_energy_time_series.copy()
    
```

```python
def peak_hour_lebeling(data_by_hour):

    labeled_data = []
    
    for data in data_by_hour:
        hour = data.get('hour')
                        
        if  (hour >= 18) & (hour <= 22):
            data['peak-hour'] = True
        else:
            data['peak-hour'] = False

        labeled_data.append(data)
    
    return labeled_data
```

```python
peak_hour_lebeling(data_by_hour)
```
