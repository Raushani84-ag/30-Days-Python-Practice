## 🔹 Task 1 — Validate Timestamp Field

```python
from datetime import datetime

def is_valid_timestamp (timestamp_str):

    if  timestamp_str is None:
        return False

    try:
        datetime.strptime(timestamp_str, "%Y-%m-%d %H:%M")
        return  True
        
    except(ValueError, TypeError):
        return False 
                
```

## 🔹 Task 2 — Validate a Single Energy Record

```python
ideal_record = {
    "site_id": "S1",
    "timestamp": "2024-01-01 08:00",
    "demand_kw": 120
}
```

```python
record_1 = {"site_id": None, "timestamp": "2024-01-01 08:00", "demand_kw": 80}
record_2 = {"site_id": "S3", "timestamp": "2024-01-01 09:00", "demand_kw": -40}
record_3 = {"site_id": "S2", "timestamp": "bad_time", "demand_kw": 60}
record_4 = {"timestamp": "2024-01-01 10:00", "demand_kw": 50}
```

```python
def validate_energy_record(record):
    if not isinstance (record , dict):
        return False , "Record must be a dictionary"
        
    site_id = record.get("site_id")
    if not  site_id  or not isinstance(site_id , str):
        return False , "Invalid or missing site_id"

    timestamp = record.get("timestamp")

    if not is_valid_timestamp(timestamp):   # Reusing Task 1 method 
        return False , "Invalid timestamp format"

    demand = record.get("demand_kw" ,0)
    if not isinstance(demand, (int, float)):
        return False , "demand_kw must be numeric"
        
    if demand < 0:
        return False, "demand_kw must be non-negetive"

    return True, None        
    
```

```python
validate_energy_record(ideal_record)
```

```python
validate_energy_record(record_1)
```

```python
validate_energy_record(record_3)
```

## 🔹 Task 3 — Validate a Batch of Records

```python
def validate_energy_batch(records):
    
    if not isinstance(records ,list):
        raise TypeError ("records must be a list")
        
    valid_records = []
    invalid_records =[]

    for record in records:
        is_valid , error = validate_energy_record(record)

        if is_valid:
            valid_records.append(record)
        else:
            invalid_records.append((record, error))
            
    return valid_records , invalid_records
      
```

```python
batch = [
    {"site_id": "S1", "timestamp": "2024-01-01 08:00", "demand_kw": 120},
    {"site_id": None, "timestamp": "2024-01-01 08:00", "demand_kw": 80},
    {"site_id": "S2", "timestamp": "bad_time", "demand_kw": 60},
    {"site_id": "S3", "timestamp": "2024-01-01 09:00", "demand_kw": -40},
]
```

```python
valid , invalid = validate_energy_batch(batch)

print("Valid : ",valid )

print("Invalid : " ,invalid )
```
