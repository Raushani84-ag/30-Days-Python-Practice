```python
def classify_sensor_readings(readings, max_irradiance = 1100):
    
    if not isinstance(readings, list):
        raise TypeError("readings must be a list")

    valid =[]
    invalid = []
    suspicious = []
    missing = []
    for data in readings:
        if data is None:
            missing.append(data)
        elif data < 0:
            invalid.append(data)
        elif data > max_irradiance:
            suspicious.append(data)
        else:
            valid.append(data)
    sensor_data_classifications = {
        'valid':valid,
        'invalid' : invalid,
        'suspicious' : suspicious,
        'missing': missing,
        'Summary': {'total data' :len(readings),
                   'valid data':len(valid),
                    'invalid data' : len(invalid),
                   'suspicious data' : len(suspicious),
                   'missing data' : len(missing)}
        }

    return sensor_data_classifications           
            
            
```

```python
def classify_single_reading(value, max_irradiance):
    if value is None:
        return "missing"
    elif value < 0:
        return "invalid"
    elif value > max_irradiance:
        return "suspicious"
    else:
        return "valid"
        
def classify_sensor_readings(readings, max_irradiance=1100):

    if not isinstance(readings, list):
        raise TypeError("readings must be a list")

    classified = {
        "valid": [],
        "invalid": [],
        "suspicious": [],
        "missing": []
    }

    for value in readings:
        label = classify_single_reading(value, max_irradiance)
        classified[label].append(value)

    return classified
    
def generate_summary(classified_data):
    return {
        "total": sum(len(v) for v in classified_data.values()),
        "valid": len(classified_data["valid"]),
        "invalid": len(classified_data["invalid"]),
        "suspicious": len(classified_data["suspicious"]),
        "missing": len(classified_data["missing"]),
    }

solar_irradiance_readings = [800, 670, -780, -200, None, 0, 1800, 670, None, 900, 1000 ]

data = classify_sensor_readings(solar_irradiance_readings, max_irradiance=1100)
summary = generate_summary(data)

output = {
    "data": data,
    "summary": summary
}

output
```
